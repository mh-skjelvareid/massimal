```python
import datetime
import shutil
from pathlib import Path

from massipipe import Pipeline, find_datasets, update_yaml
```

```python
# Paths
dataset_base_dirs = [
    "/media/mha114/Massimal2/seabee-minio/bodo/juvika/aerial/hsi/20220624",
    "/media/mha114/Massimal2/seabee-minio/bodo/sandsund/aerial/hsi/20210602/massimal_bodo_sandsund_202106021653-georef_hsi",
    "/media/mha114/Massimal2/seabee-minio/larvik/olbergholmen/aerial/hsi/20230830/massimal_larvik_olbergholmen_202308300959-south1_hsi",
    "/media/mha114/Massimal2/seabee-minio/larvik/kongsbakkebukta/aerial/hsi/20230830",
    "/media/mha114/Massimal2/seabee-minio/smola/maholmen/aerial/hsi/20230621",
    "/media/mha114/Massimal2/seabee-minio/vega/sola/aerial/hsi/20220823",
]
```

```python
dataset_dirs = []
for dataset_base_dir in dataset_base_dirs:
    dataset_dirs += find_datasets(Path(dataset_base_dir))
print(f"Found {len(dataset_dirs)} datasets")
```

```python
# Update YAML files to create radiance_gc and reflectance_gc
for dataset_dir in dataset_dirs:
    print(dataset_dir)
    try:
        pipeline = Pipeline(dataset_dir)
        orig_config_path = pipeline.config_file_path
        update_yaml(
            orig_config_path,
            ["massipipe_options", "radiance_gc", "create"],
            value=True,
            key_must_exist=False,
        )
        update_yaml(
            orig_config_path,
            ["massipipe_options", "radiance_gc_rgb", "create"],
            value=True,
            key_must_exist=False,
        )
        update_yaml(
            orig_config_path,
            ["massipipe_options", "reflectance", "create"],
            value=True,
            key_must_exist=False,
        )
        update_yaml(
            orig_config_path,
            ["massipipe_options", "reflectance_gc", "create"],
            value=True,
            key_must_exist=False,
        )
        # Disable RGB for reflectance, save space
        update_yaml(
            orig_config_path,
            ["massipipe_options", "reflectance_gc_rgb", "create"],
            value=False,
            key_must_exist=False,
        )

    except Exception as e:
        print(f"Error while processing {dataset_dir}")
        print(e)
```

```python
n_images = 0
for d in dataset_dirs:
    pipeline = Pipeline(d)
    n_images += len(pipeline.base_file_names)

print(f"Total number of images to process: {n_images}")
```
