# Create YAML file backups

```python
from pathlib import Path  
from massipipe.pipeline import Pipeline, find_datasets
import massipipe.config
import yaml
from rich import print
import shutil
import datetime
```

```python
base_dir = '/home/mha114/data/seabee-minio'
# base_dir = '/home/mha114/data/seabee-minio/bodo'
# base_dir = '/home/mha114/data/seabee-minio/larvik'
# base_dir = '/home/mha114/data/seabee-minio/smola'
```

```python
# timestamp = datetime.datetime.now().strftime("%Y%m%d%H%M")
# print(timestamp)
```

```python
for dataset_dir in find_datasets(Path(base_dir)):
    print(dataset_dir)
    try:
        pipeline = Pipeline(dataset_dir)
        orig_config_path = pipeline.config_file_path
        timestamp = datetime.datetime.now().strftime("%Y%m%d%H%M")
        backup_config_path = (orig_config_path.parent / 
                              ('config_backup_' + timestamp + ''.join(orig_config_path.suffixes)))
        backup_dest = shutil.copy(orig_config_path,backup_config_path)
        print(f"File copied to {backup_dest}")
        # print(orig_config_path)
        # print(backup_config_path)
        
    except Exception as e:
        print(f"Error while processing {dataset_dir}")
        print(e)
```
