```python
from pathlib import Path  
from massipipe.pipeline import PipelineProcessor, find_datasets
import massipipe.config
import yaml
from rich import print
```

```python
# Parameters
# base_dir = '/home/mha114/data/seabee-minio'
# base_dir = '/home/mha114/data/seabee-minio/larvik'
base_dir = '/home/mha114/data/seabee-minio/smola'
```

```python
def update_dict(yaml_dict:dict, keys:list[str], value,key_must_exist):
    for key in keys[0:-1]:
        yaml_dict = yaml_dict[key]
    if key_must_exist:
        if keys[-1] in yaml_dict: 
            yaml_dict[keys[-1]] = value
        else:
            print(f"{keys[-1]} not in yaml_dict")
    else:
        yaml_dict[keys[-1]] = value

def update_yaml(yaml_file,keys,value,key_must_exist):
    yaml_dict = massipipe.config.read_config(yaml_file)
    update_dict(yaml_dict,keys,value,key_must_exist)
    massipipe.config.write_config(yaml_dict,yaml_file)
    
def update_yaml_in_datasets(base_dir,keys,value,key_must_exist=True):
    for dataset_dir in find_datasets(Path(base_dir)):
        print(dataset_dir)
        try:
            pipeline = PipelineProcessor(dataset_dir)
            update_yaml(pipeline.config_file_path,keys,value,key_must_exist)
        except Exception as e:
            print(f"Error while processing {dataset_dir}")
            print(e)
                        
```

```python
update_yaml_in_datasets(
    base_dir,
    keys=['massipipe_options','radiance_gc_rgb','overwrite'],
    value=True,
    key_must_exist=False)
```
