```python
from pathlib import Path
from massipipe import find_datasets
import shutil
```

```python
base_dir = '/home/mha114/data/seabee-minio'
```

```python
dataset_dirs = find_datasets(base_dir)
```

```python
for dataset_dir in dataset_dirs:
    print("\n----\n")
    print(dataset_dir)
    orthomosaic_dir = (dataset_dir / "orthomosaic")

    if orthomosaic_dir.exists():
        old_mosaic_file_list = sorted(orthomosaic_dir.glob("*.tiff"))
        if len(old_mosaic_file_list) == 0:
            print("NO MOSAICS FOUND")
        elif len(old_mosaic_file_list) > 1:
            print("MORE THAN ONE MOSAIC")
        else:
            old_mosaic_path = old_mosaic_file_list[0]
            new_mosaic_path = old_mosaic_path.parent / f"{old_mosaic_path.stem}_orthophoto.original.tif"
            print(f"Renaming \n  {old_mosaic_path} to \n  {new_mosaic_path}")
            shutil.move(old_mosaic_path,new_mosaic_path)
            
            new_mosaic_dir = dataset_dir / "orthophoto"
            print(f"Renaming \n  {orthomosaic_dir} to \n  {new_mosaic_dir}")
            shutil.move(orthomosaic_dir,new_mosaic_dir)

        
```
