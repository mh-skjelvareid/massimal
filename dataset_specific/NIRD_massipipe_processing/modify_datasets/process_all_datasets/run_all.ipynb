# Run all steps for all(!) datasets

```python
from pathlib import Path  
from massipipe import Pipeline, find_datasets
from rich import print
```

```python
# Parameters
base_dir = '/home/mha114/data/seabee-minio'
```

```python
for dataset_dir in find_datasets(Path(base_dir)):
    print(dataset_dir.name)
```

```python
for dataset_dir in find_datasets(Path(base_dir)):
    try:
        pipeline = Pipeline(dataset_dir)
        pipeline.delete_existing_products(delete_quicklook=False)
        pipeline.run()
    except Exception:
        print(f"Error while processing {dataset_dir}")
```
