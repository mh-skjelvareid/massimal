```python
import massipipe.pipeline
from pathlib import Path
import shutil
```

```python
base_dir = Path('/home/mha114/data/seabee-minio/')
```

```python
radiance_dirs = sorted(d for d in base_dir.rglob("**/0_radiance"))
data_dirs = [d.parent for d in radiance_dirs]
```

```python
for data_dir in data_dirs:
    print(data_dir)
```

```python
for rad_dir, data_dir in zip(radiance_dirs, data_dirs):
    raw_dir = data_dir / "0_raw"
    raw_dir.mkdir(exist_ok=True)
    
    image_dirs = sorted(md for md in rad_dir.glob("*"))
    for image_dir in image_dirs:
        # print(image_dir)
        
        raw_im_dir = raw_dir / image_dir.name
        raw_im_dir.mkdir(exist_ok=True)
        copy_file_list = image_dir.glob("*")
        copy_file_list = [f for f in copy_file_list if "radiance" not in f.name]
        
        # for copy_file in copy_file_list:
        #     print(f"   {copy_file}")
        for copy_file in copy_file_list:
            shutil.copy(copy_file,raw_im_dir)
            
        radiance_header = list(image_dir.glob("*radiance.bil.hdr"))[0]
        raw_header_name = radiance_header.name.replace("-radiance","")
        shutil.copy(radiance_header,raw_im_dir/raw_header_name)
```

```python
for old_rad_dir, data_dir in zip(radiance_dirs, data_dirs):
    raw_dir = data_dir / "0_raw"
    new_rad_dir = data_dir / "1a_radiance"
    new_rad_dir.mkdir(exist_ok=True)
    
    pipeline = massipipe.pipeline.PipelineProcessor(data_dir)
    raw_paths = raw_dir.rglob("*.bil.hdr")
    raw_paths = sorted(raw_paths,key=pipeline._get_image_number)
    base_file_names = [f"{pipeline.dataset_base_name}_{i:03d}" for i in range(len(raw_paths))]
    
    for raw_path,base_file_name in zip(raw_paths,base_file_names):
        orig_rad_path = old_rad_dir / (raw_path.parent.name) / (raw_path.name.split(".")[0] + "-radiance.bil")
        new_rad_path = new_rad_dir / (base_file_name+"_radiance.bip")
        
        print(f"Copying{orig_rad_path} to {new_rad_path}")
        shutil.copy(orig_rad_path,new_rad_path)
        shutil.copy(orig_rad_path.parent / (orig_rad_path.name + '.hdr'),
                    new_rad_path.parent / (new_rad_path.name + '.hdr'))
```
